Introducción
La regresión lineal es una técnica estadística ampliamente utilizada para identificar tendencias y relaciones entre variables. En este proyecto se emplea esta metodología para analizar la evolución de las emisiones de dióxido de carbono (CO₂) en Colombia durante el periodo comprendido entre 1990 y 2025.

A partir de los datos previamente limpiados y preparados por el equipo de trabajo, se construye un modelo de regresión lineal con el objetivo de estudiar la relación existente entre los años analizados y las emisiones de CO₂ registradas en el país. Posteriormente, se interpretan los resultados obtenidos para comprender el comportamiento de esta variable ambiental a lo largo del tiempo.

Objetivo
Desarrollar un modelo de regresión lineal que permita analizar la evolución de las emisiones de CO₂ en Colombia e interpretar las tendencias observadas durante el periodo de estudio.

Carga de datos
En esta sección se cargan los datos previamente depurados por el equipo de trabajo. Estos datos contienen la información histórica de emisiones de CO₂ en Colombia desde 1990 hasta 2025.

```python
import pandas as pd

# Cargar datos limpios
df = pd.read_csv("datos_limpios_colombia.csv")

print("✅ Datos cargados correctamente")
print(f"Filas: {df.shape[0]}")
print(f"Columnas: {df.shape[1]}")

df.head()
```

Observación
Los datos cargados contienen las variables país, código, año y emisiones de CO₂, las cuales serán utilizadas para construir el modelo de regresión lineal.

Preparación de variables
Para construir el modelo de regresión lineal se utilizará el año como variable independiente (X) y las emisiones de CO₂ como variable dependiente (y). Estas variables permitirán analizar cómo han cambiado las emisiones a lo largo del tiempo y construir un modelo predictivo.

```python
# Variable independiente (Año)
X = df[['anio']]

# Variable dependiente (Emisiones de CO₂)
y = df['emisiones_co2']

print("Primeros valores de X:")
print(X.head())

print("\nPrimeros valores de y:")
print(y.head())
```

Interpretación
La variable X representa el año de registro de los datos, mientras que la variable y representa las emisiones de CO₂ correspondientes a cada año. Estas dos variables serán utilizadas para entrenar el modelo de regresión lineal y analizar la tendencia de las emisiones en Colombia.

Entrenamiento del modelo
Una vez definidas las variables independientes y dependientes, se procede a entrenar un modelo de regresión lineal. Este modelo permitirá identificar la tendencia de las emisiones de CO₂ en Colombia a lo largo del tiempo.

```python
from sklearn.linear_model import LinearRegression

# Crear modelo
modelo = LinearRegression()

# Entrenar modelo
modelo.fit(X, y)

print("✅ Modelo entrenado correctamente")
```

Interpretación
El modelo de regresión lineal ha sido entrenado utilizando los datos históricos disponibles. A partir de este punto será posible generar predicciones y analizar la tendencia general de las emisiones de CO₂ en Colombia.

Generación de predicciones y visualización gráfica
Una vez entrenado el modelo, se generan las predicciones correspondientes para cada año y se construye una representación gráfica que permite comparar los datos observados con la tendencia estimada por la regresión lineal.

```python
# Generar predicciones
predicciones = modelo.predict(X)

print("Primeras predicciones:")
print(predicciones[:5])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.scatter(X, y,
            color='blue',
            label='Datos reales')

plt.plot(X,
         predicciones,
         color='red',
         linewidth=3,
         label='Regresión lineal')

plt.title('Regresión Lineal de Emisiones de CO₂ en Colombia')
plt.xlabel('Año')
plt.ylabel('Emisiones de CO₂')
plt.legend()
plt.grid(True)

plt.show()
```

Se generan predicciones y gráfica de regresión lineal

Evaluación e interpretación del modelo
Para evaluar el desempeño del modelo se utiliza el coeficiente de determinación R². Este indicador permite medir qué tan bien la regresión lineal explica la variabilidad observada en los datos.

Posteriormente se realiza una interpretación de los resultados obtenidos y de la tendencia observada en las emisiones de CO₂.

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, predicciones)

print("Coeficiente R²:", r2)
```

Interpretación de resultados
El modelo de regresión lineal permite observar la tendencia general de las emisiones de CO₂ en Colombia durante el periodo analizado.

La línea de regresión resume el comportamiento histórico de los datos y permite identificar la dirección general de cambio en las emisiones registradas.

El valor de R² indica el nivel de ajuste del modelo respecto a los datos reales y ayuda a determinar la capacidad explicativa de la regresión lineal.
