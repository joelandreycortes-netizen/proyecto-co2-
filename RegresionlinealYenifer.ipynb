Introducción
La regresión lineal es una técnica estadística ampliamente utilizada para identificar tendencias y relaciones entre variables. En este proyecto se emplea esta metodología para analizar la evolución de las emisiones de dióxido de carbono (CO₂) en Colombia durante el periodo comprendido entre 1990 y 2025.

A partir de los datos previamente limpiados y preparados por el equipo de trabajo, se construye un modelo de regresión lineal con el objetivo de estudiar la relación existente entre los años analizados y las emisiones de CO₂ registradas en el país. Posteriormente, se interpretan los resultados obtenidos para comprender el comportamiento de esta variable ambiental a lo largo del tiempo.

Objetivo
Desarrollar un modelo de regresión lineal que permita analizar la evolución de las emisiones de CO₂ en Colombia e interpretar las tendencias observadas durante el periodo de estudio.
